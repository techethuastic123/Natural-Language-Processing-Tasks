## Task C1: Speed test

```python
# multiply every item by 2 with a list comprehension and with NumPy, 5 runs each, averaged
sizes = [1_000, 100_000, 1_000_000]
rows = []
for n in sizes:
    py_list = list(range(n))
    np_arr = np.arange(n)

    t_list = []
    t_np = []
    for _ in range(5):
        s = time.perf_counter()
        r1 = [x * 2 for x in py_list]
        t_list.append(time.perf_counter() - s)

        s = time.perf_counter()
        r2 = np_arr * 2
        t_np.append(time.perf_counter() - s)

    avg_l, avg_n = sum(t_list) / 5, sum(t_np) / 5
    rows.append({"size": n, "list_time_s": avg_l, "numpy_time_s": avg_n, "speed_up": avg_l / avg_n})

speed = pd.DataFrame(rows)
pd.set_option("display.float_format", lambda v: f"{v:.8f}")
print(speed.to_string(index=False))
pd.reset_option("display.float_format")
```

**Why the gap changes with size (my run: about 2.8x at 1,000, about 20.6x at 100,000, about 15x at 1,000,000):**  
A Python list holds separate objects, so the loop checks the type and creates a new int object for every item. NumPy stores plain numbers in one block of memory and multiplies them in fast compiled code. At 1,000 items the work is so small that NumPy's fixed start-up cost eats much of its advantage, so the speed-up is small. At 100,000 that fixed cost is spread over many items and the gap is largest. At 1,000,000 the arrays no longer fit in the CPU cache, so memory speed limits NumPy a bit and the ratio dips. Exact numbers vary with my machine and each run.

## Task C4: Mini text-analysis pipeline

### C4.1 Sentences and features

```python
# 12 sentences of my own (4+ words each, 4 contain digits, one has a question mark)
sentences = [
    "The NLP lab at our campus starts at 9 in the morning.",
    "I study natural language processing with Python and NumPy.",
    "Our lab has 30 computers and a fast internet connection.",
    "The teacher explains the lab tasks very clearly.",
    "Pandas makes it easy to clean messy data.",
    "I finished 3 of the 8 NumPy tasks yesterday.",
    "Why does the loop run slower than the NumPy array?",
    "The deadline for the assignment is the 17th of October.",
    "Students must write the code in their own words.",
    "A bag of words matrix counts every word in a sentence.",
    "Cosine similarity compares two sentences using the angle between them.",
    "My friends and I practice the lab tasks every evening.",
]
text_df = pd.DataFrame({"text": sentences})

# helper: remove punctuation so 'array?' and 'array' count as the same word
PUNCT = ".,?!;:\"'()-"
def words_of(t):
    return t.translate(str.maketrans("", "", PUNCT)).split()

text_df["word_count"] = text_df["text"].apply(lambda t: len(words_of(t)))
text_df["char_count"] = text_df["text"].str.len()
text_df["avg_word_length"] = text_df["text"].apply(
    lambda t: round(np.mean([len(w) for w in words_of(t)]), 2))
text_df["has_digit"] = text_df["text"].apply(lambda t: any(ch.isdigit() for ch in t))
text_df["longest_word"] = text_df["text"].apply(lambda t: max(words_of(t), key=len))

pd.set_option("display.width", 200); pd.set_option("display.max_colwidth", 60)
print(text_df.to_string())
print("\nSentences with a digit:", text_df["has_digit"].sum())
```

### C4.2 Sorting and groupby

```python
# sort by word_count, biggest first, then keep the top 3
print(text_df.sort_values("word_count", ascending=False).head(3)[["text", "word_count"]].to_string())

# groupby splits the rows into groups by has_digit, then mean() is computed inside each group
print()
print(text_df.groupby("has_digit")["word_count"].mean())
```

**What groupby does:** `groupby("has_digit")` splits the table into groups (one group of rows with a digit, one without), then an aggregation such as `.mean()` is applied to each group separately and the answers are combined into one small result. It is like sorting the sentences into two piles and averaging each pile.

### C4.3 Z-score with NumPy vs pandas

```python
# word_count -> NumPy array, then z = (x - mean) / std computed by hand
wc = text_df["word_count"].to_numpy()
z_np = (wc - wc.mean()) / wc.std()                       # np.std uses ddof=0 (divide by n)
z_pd = (text_df["word_count"] - text_df["word_count"].mean()) / text_df["word_count"].std()  # pandas uses ddof=1 (n-1)

print("NumPy std :", round(wc.std(), 4), "  pandas std:", round(text_df["word_count"].std(), 4))
cmp = pd.DataFrame({"word_count": wc, "z_numpy": np.round(z_np, 4), "z_pandas": np.round(z_pd, 4)})
print(cmp.to_string())

# the pandas result can be reproduced in NumPy with ddof=1
print("Same as pandas when ddof=1:", np.allclose(z_pd.to_numpy(), (wc - wc.mean()) / wc.std(ddof=1)))
```

**Why they differ:** NumPy's `std` uses `ddof=0` (divides by *n*, the population standard deviation), while pandas' `.std()` uses `ddof=1` (divides by *n − 1*, the sample standard deviation). A bigger denominator inside the square root gives a slightly larger std, so the pandas z-scores are slightly smaller in size. Setting `ddof=1` in NumPy makes them identical.

### C4.4 Bag-of-words and cosine similarity

```python
# vocabulary = all unique lowercase words from the 12 sentences (sorted so the column order is fixed)
tokenised = [[w.lower() for w in words_of(t)] for t in sentences]
vocab = sorted(set(w for sent in tokenised for w in sent))
index = {w: i for i, w in enumerate(vocab)}

# bag-of-words matrix: row = sentence, column = word, value = how many times it appears
bow = np.zeros((len(sentences), len(vocab)), dtype=int)
for i, sent in enumerate(tokenised):
    for w in sent:
        bow[i, index[w]] += 1

print("Vocabulary size:", len(vocab))
print("BoW shape:", bow.shape)

# cosine similarity = dot(a, b) / (|a| * |b|)
def cosine(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

print("\nCosine similarity of sentence 1 with the others:")
for j in range(1, len(sentences)):
    print(f"  S1 vs S{j+1}: {cosine(bow[0], bow[j]):.4f}")

# most similar pair over all pairs (i < j)
best, pair = -1, None
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        c = cosine(bow[i], bow[j])
        if c > best:
            best, pair = c, (i, j)
print(f"\nMost similar pair: S{pair[0]+1} and S{pair[1]+1} (cosine = {best:.4f})")
print("S%d: %s" % (pair[0]+1, sentences[pair[0]]))
print("S%d: %s" % (pair[1]+1, sentences[pair[1]]))
```

**Meaning of the values:** each sentence is a vector of word counts and cosine similarity measures the angle between two such vectors.
- **Similarity 0:** the sentences share no words at all (the vectors are at 90°, completely different in this representation).
- **Similarity 1:** the vectors point in exactly the same direction, so both sentences use the same words in the same proportions (e.g. identical sentences, or one repeated twice).
- Values in between show partial overlap. Note it only counts words, so word order and meaning are ignored.

## Task C5: Short answers

**1. Why do `[1, 2, 3] * 2` and `np.array([1, 2, 3]) * 2` differ?**
For a Python list, `*` means repetition, so the list is repeated and we get `[1, 2, 3, 1, 2, 3]`. A NumPy array does maths on each element, so we get `[2 4 6]`. Same symbol, different meaning, because the two types define `*` differently.

**2. What do `a[0]`, `a[[0, 2]]` and `a[:, 0]` return?**
`a[0]` is the first row (a 1-D array). `a[[0, 2]]` uses a list of indexes (fancy indexing) and returns a new array with row 0 and row 2. `a[:, 0]` means all rows, column 0, so it returns the first column.

**3. axis=0 vs axis=1 with a table of student marks.**
Say each row is a student and each column is a subject. `axis=0` goes down the rows and collapses them, so `sum(axis=0)` gives one total per *subject* (all students added together). `axis=1` goes across the columns, so `sum(axis=1)` gives one total per *student* (all of his subjects added). The axis named is the one that disappears.

**4. Why does `fillna(0)` change the mean, and when is the mean better?**
Filling with 0 adds fake very low ratings, which pulls the average down, whereas the original mean ignored the missing ones. Filling with the column mean keeps the average the same. The mean is better when the value is a measurement with a typical level (ratings, prices), because 0 is not a real possible value there. 0 is fine when missing really means "none", like units sold when nothing was sold.
